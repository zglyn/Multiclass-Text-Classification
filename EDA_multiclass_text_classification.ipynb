Import the csv file from s3

```python
import pandas as pd
s3_path = 's3://hf-course-multiclass-text-classification-bucket/training_data/'

original_df = pd.read_csv(s3_path, sep='\t', names = ["ID", "TITLE", "URL", "PUBLISHER", "CATEGORY", "STORY", "HOSTNAME", "TIMESTAMP"])
# FORMAT: ID 	 TITLE 	 URL 	 PUBLISHER 	 CATEGORY 	 STORY 	 HOSTNAME 	 TIMESTAMP
original_df.head()
```

Only need the Title and Category 
Then, change category names into the full name
* B = business, m = medical, t = technology, e = entertainment 

```python
def update_category(category):
    my_dict = {
        'b': 'business',
        't': 'science',  
        'm': 'health',
        'e': 'entertainment',
    }
    return my_dict[category]
```

```python
cleaned = original_df.copy() 
cleaned = cleaned[['TITLE', 'CATEGORY']]
cleaned['CATEGORY'] = cleaned['CATEGORY'].apply(update_category)
```
